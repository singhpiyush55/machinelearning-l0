```python
from sklearn.datasets import load_digits
```

```python
type(load_digits)
```

| Component    | Description                    |
| ------------ | ------------------------------ |
| Samples      | **1,797 images**               |
| Classes      | **10 digits (0–9)**            |
| Image size   | **8 × 8 pixels**               |
| Features     | **64 features per image**      |
| Pixel values | **0–16 (grayscale intensity)** |

Each image is flattened into a row of 64 numbers.

```python
import matplotlib.pyplot as plt

plt.imshow(digits.images[0], cmap='grey')
plt.title(f"Label: {digits.target[0]}")
plt.axis('off')
plt.show()
```

```python
# Now training
from sklearn.model_selection import train_test_split
```

```python
# for x use digit.data not digits.images.
X_train, X_test, y_train, y_test = train_test_split(digits.data, digits.target, test_size=0.2, random_state = 42)
```

```python
print(X_train.shape, X_test.shape)
```

```python
y_test
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression()
```

```python
model.fit(X_train, y_train)
```

```python
model.score(X_test, y_test)
```

```python
from sklearn.metrics import confusion_matrix
```

```python
y_pred = model.predict(X_test)
```

```python
cm = confusion_matrix(y_test, y_pred)
cm
```

```python
import seaborn as sn

sn.heatmap(cm, annot=True)
plt.xlabel('Predicted')
plt.ylabel('Truth')
```
