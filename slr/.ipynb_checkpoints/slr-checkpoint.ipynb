```python
import pandas as pd
```

```python
df = pd.read_csv('./homeprices_slr_l2.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
import matplotlib.pyplot as plt
```

```python
# Plotting to see the relationship
plt.scatter(df.area, df.price)
plt.xlabel('Area')
plt.ylabel('Price')
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression();
```

```python
#Training
model.fit(df[['area']], df.price)
```

```python
model.predict([[3700]])
```

```python
model.coef_
```

```python
model.intercept_
```

```python
new_test_df = pd.read_csv('./areas_slr_l2.csv')
```

```python
new_test_df.head()
```

```python
# prediction on values from files.
results = model.predict(new_test_df)
```

```python
results
```

```python
new_test_df['results'] = results
```

```python
new_test_df.head()
```

```python
# exporting the df with predicted values. 
new_test_df.to_csv('./areas_slr_l2.csv')
```
