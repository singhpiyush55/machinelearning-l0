```python
# One-Hot encoding.
```

```python
import pandas as pd
```

```python
df = pd.read_csv('./homeprices_encoding_l6.csv')
```

```python
df
```

```python
dummies = pd.get_dummies(df.town, dtype='int')
```

```python
dummies
```

```python
merged = pd.concat([df, dummies], axis='columns')
```

```python
merged
```

```python
new_df = merged.drop(['town'], axis='columns')
```

```python
new_df
```

```python
# We need to drop one of the dummy variables, to avoid dummy variable trap.
final_df = new_df.drop(['robinsville'], axis='columns')
```

```python
final_df
```

```python
x = final_df.drop(['price'], axis='columns')
```

```python
x
```

```python
y = final_df.price
```

```python
y
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
model.fit(x,y)
```

```python
import joblib
```

```python
joblib.dump(model, 'mlr_encoded')
```
